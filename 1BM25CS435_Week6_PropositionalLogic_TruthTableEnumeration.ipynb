```python
from itertools import product


def implies(a, b):
    return (not a) or b


def PL_TRUE_KB(model):

    p = model['P']
    q = model['Q']
    r = model['R']

    s1 = implies(q, p)
    s2 = implies(p, not q)
    s3 = q or r

    return s1 and s2 and s3


def PL_TRUE_QUERY(alpha, model):

    p = model['P']
    q = model['Q']
    r = model['R']

    if alpha == 'R':
        return r

    elif alpha == 'R -> P':
        return implies(r, p)

    elif alpha == 'Q -> R':
        return implies(q, r)


def print_truth_table():

    print("\nCOMPLETE TRUTH TABLE\n")

    print("P Q R | ~Q | Q->P | P->~Q | QvR | KB")
    print("-" * 48)

    models = []

    for p, q, r in product([True, False], repeat=3):

        model = {'P': p, 'Q': q, 'R': r}

        s1 = implies(q, p)
        s2 = implies(p, not q)
        s3 = q or r

        kb = s1 and s2 and s3
        values = [p, q, r, not q, s1, s2, s3, kb]

        row = [
            'T' if value else 'F'
            for value in values
        ]

        print(
            f"{row[0]} {row[1]} {row[2]} | "
            f" {row[3]} |  {row[4]}  |"
            f"   {row[5]}  |  {row[6]}  | {row[7]}"
        )

        if kb:
            models.append(model)

    print("\nModels satisfying KB:")

    for model in models:
        print(model)

    return models


def TT_CHECK_ALL(KB, alpha, symbols, model):
    if len(symbols) == 0:

        if KB(model):
            return PL_TRUE_QUERY(alpha, model)

        else:
            return True

    else:
        P = symbols[0]
        rest = symbols[1:]
        true_result = TT_CHECK_ALL(
            KB, alpha, rest,
            {**model, P: True}
        )
        false_result = TT_CHECK_ALL(
            KB, alpha, rest,
            {**model, P: False}
        )
        return true_result and false_result


def TT_ENTAILS(KB, alpha):

    symbols = ['P', 'Q', 'R']

    model = {}

    return TT_CHECK_ALL(
        KB, alpha, symbols, model
    )


models = print_truth_table()

print("\nENTAILMENT RESULTS\n")

queries = ['R', 'R -> P', 'Q -> R']

for query in queries:

    result = TT_ENTAILS(PL_TRUE_KB, query)

    print("Query:", query)

    if result:
        print("KB entails", query)
    else:
        print("KB does not entail", query)

    print("---------------")
```
