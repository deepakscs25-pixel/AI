```python
def is_variable(x):
    return isinstance(x, str) and len(x) == 1 and x.islower()


def occurs(var, expr):
    if var == expr:
        return True

    if isinstance(expr, tuple):
        return any(occurs(var, a) for a in expr[1])

    return False


def apply_substitution(expr, subst):
    while isinstance(expr, str) and expr in subst:
        expr = subst[expr]

    if isinstance(expr, tuple):
        return (
            expr[0],
            tuple(apply_substitution(a, subst)
                  for a in expr[1])
        )

    return expr


def compose(s, subst):
    updated = {
        key: apply_substitution(value, s)
        for key, value in subst.items()
    }

    updated.update(s)
    return updated


def unify(x, y):
    if x == y:
        return {}

    if is_variable(x):
        if occurs(x, y):
            return "FAILURE"
        return {x: y}

    if is_variable(y):
        if occurs(y, x):
            return "FAILURE"
        return {y: x}

    if not isinstance(x, tuple) or not isinstance(y, tuple):
        return "FAILURE"

    if x[0] != y[0] or len(x[1]) != len(y[1]):
        return "FAILURE"

    subst = {}

    for a, b in zip(x[1], y[1]):
        a = apply_substitution(a, subst)
        b = apply_substitution(b, subst)

        s = unify(a, b)

        if s == "FAILURE":
            return "FAILURE"

        subst = compose(s, subst)

    return subst


def parse_expression(expr):
    expr = expr.replace(" ", "")

    predicate = expr[:expr.index("(")]

    arguments = expr[
        expr.index("(") + 1:expr.rindex(")")
    ]

    args = tuple(arguments.split(","))

    return (predicate, args)


print("FIRST ORDER LOGIC - UNIFICATION")

x = input("Enter first expression: ")
y = input("Enter second expression: ")

x = parse_expression(x)
y = parse_expression(y)

result = unify(x, y)

if result == "FAILURE":
    print("\nUnification Failed")
else:
    unified = apply_substitution(x, result)

    output = unified[0] + "(" + ", ".join(unified[1]) + ")"

    print("\nUnified Expression:")
    print(output)
```
