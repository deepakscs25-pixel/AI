```python
def negate(literal):
    if literal.startswith("~"):
        return literal[1:]
    return "~" + literal


def resolution(KB, query):
    clauses = set()

    for clause in KB:
        clauses.add(frozenset(clause))

    clauses.add(frozenset([negate(query)]))

    while True:
        new = set()
        clause_list = list(clauses)

        for i in range(len(clause_list)):
            for j in range(i + 1, len(clause_list)):

                C1 = clause_list[i]
                C2 = clause_list[j]

                for literal in C1:
                    opposite = negate(literal)

                    if opposite in C2:
                        resolvent = (C1 - {literal}) | (C2 - {opposite})

                        if len(resolvent) == 0:
                            return True

                        if not any(negate(x) in resolvent for x in resolvent):
                            if resolvent not in clauses:
                                new.add(resolvent)

        if len(new) == 0:
            return False

        clauses.update(new)
        print("New clauses:", new)


KB = [
    ["~Food(Peanuts)", "Likes(John,Peanuts)"],
    ["~Eats(Anil,Peanuts)", "Killed(Anil)", "Food(Peanuts)"],
    ["Eats(Anil,Peanuts)"],
    ["Alive(Anil)"],
    ["~Alive(Anil)", "~Killed(Anil)"]
]

query = "Likes(John,Peanuts)"

if resolution(KB, query):
    print("\nJohn likes Peanuts - PROVED")
else:
    print("\nCannot prove the query")
```
