```python
KB = {
    "American(Robert)",
    "Missile(T1)",
    "Owns(A,T1)",
    "Enemy(A,America)"
}


def forward_chaining(KB, query):

    while True:
        new = set()
        if "Missile(T1)" in KB and "Owns(A,T1)" in KB:
            new.add("Sells(Robert,T1,A)")

        if "Missile(T1)" in KB:
            new.add("Weapon(T1)")


        if "Enemy(A,America)" in KB:
            new.add("Hostile(A)")

        if ("American(Robert)" in KB
            and "Weapon(T1)" in KB
            and "Sells(Robert,T1,A)" in KB
            and "Hostile(A)" in KB):

            new.add("Criminal(Robert)")

        if query in KB or query in new:
            return True

        new = new - KB

        if len(new) == 0:
            return False

        KB.update(new)

        print("New facts:", new)


query = "Criminal(Robert)"

if forward_chaining(KB, query):
    print("\nRobert is criminal")
else:
    print("\nCannot prove Robert is criminal")
```
