```python
import heapq

def misplaced(state, goal):
    count = 0

    for i in range(9):
        if state[i] != 0 and state[i] != goal[i]:
            count += 1

    return count


def neighbors(state):
    result = []

    pos = state.index(0)
    row = pos // 3
    col = pos % 3

    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in moves:

        r = row + dr
        c = col + dc

        if 0 <= r < 3 and 0 <= c < 3:

            new_pos = r * 3 + c

            new_state = list(state)

            new_state[pos], new_state[new_pos] = \
                new_state[new_pos], new_state[pos]

            result.append(tuple(new_state))

    return result


def astar(start, goal):

    pq = []

    h = misplaced(start, goal)

    heapq.heappush(pq, (h, 0, start, []))

    visited = set()

    while pq:

        f, g, state, path = heapq.heappop(pq)

        if state in visited:
            continue

        visited.add(state)

        path = path + [state]

        if state == goal:
            return path

        for next_state in neighbors(state):

            if next_state not in visited:

                new_g = g + 1
                h = misplaced(next_state, goal)
                new_f = new_g + h

                heapq.heappush(
                    pq,
                    (new_f, new_g, next_state, path)
                )

    return None


def print_board(state):

    for i in range(0, 9, 3):
        print(state[i:i+3])

    print()


start = (
    2, 8, 3,
    1, 6, 4,
    7, 0, 5
)

goal = (
    1, 2, 3,
    8, 0, 4,
    7, 6, 5
)

solution = astar(start, goal)

if solution:

    print("Solution found!")
    print("Moves:", len(solution) - 1)

    for i, state in enumerate(solution):
        print("Step", i)
        print_board(state)

else:
    print("No solution")
```
