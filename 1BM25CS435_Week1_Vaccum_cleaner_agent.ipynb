```python
room_A = "Dirty"
room_B = "Dirty"

position = "A"
iteration = 1

while room_A == "Dirty" or room_B == "Dirty":

    print("\nIteration:", iteration)
    print("Vacuum position:", position)
    print("Room A:", room_A)
    print("Room B:", room_B)

    if position == "A":
        if room_A == "Dirty":
            print("Action: Clean Room A")
            room_A = "Clean"
        else:
            print("Action: Move to Room B")
            position = "B"

    elif position == "B":
        if room_B == "Dirty":
            print("Action: Clean Room B")
            room_B = "Clean"
        else:
            print("Action: Move to Room A")
            position = "A"

    iteration += 1

print("\nBoth rooms are clean!")
print("Final Status:")
print("Room A:", room_A)
print("Room B:", room_B)
```
